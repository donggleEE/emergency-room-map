```python
import pandas as pd

df = pd.read_csv(
    "data/hospital_er_preprocessed_wide.csv",
    encoding="utf-8-sig",
    dtype={
        "hpid": str,
        "realtime_updated_at": str,
        "collected_at": str,
    }
)

check_cols = [
    "avail_er_general_beds",
    "avail_operating_rooms",
    "avail_icu_general",
    "avail_inpatient_general_beds",
    "delivery_room_status",
    "available_ct_yn",
    "available_mri_yn",
    "realtime_updated_at",
]

existing_cols = [col for col in check_cols if col in df.columns]

df["has_realtime_info"] = df[existing_cols].notna().any(axis=1)

print("전체 병원 수:", len(df))
print("실시간 정보 있는 병원 수:", df["has_realtime_info"].sum())
print("실시간 정보 없는 병원 수:", (~df["has_realtime_info"]).sum())

no_info = df[~df["has_realtime_info"]][
    [
        "hpid",
        "hospital_name",
        "sido",
        "sigungu",
        "dutyAddr",
        "emergency_tel",
    ]
]

no_info.head(30)
```
